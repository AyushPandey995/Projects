```python
import numpy as np
```

```python
#Creating Income Categories
df['Income_Category'] = pd.cut(df['median_income'], bins = [0, 1.5, 3, 4.5, 6, np.inf], labels = [1, 2, 3, 4, 5])
```

```python
df
```

```python
import matplotlib.pyplot as plt
```

```python
df['Income_Category'].value_counts().sort_index().plot.bar(rot = 0, grid = True)
plt.title('Income Category Distribution')
plt.xlabel('Income Categories')
plt.ylabel('No. of Instances')
```

```python
from sklearn.model_selection import StratifiedShuffleSplit
```

```python
split = StratifiedShuffleSplit(n_splits = 1, test_size = 0.2, random_state = 42)
split.split(df, df['Income_Category'])
```

```python
for train_index, test_index in split.split(df, df['Income_Category']):
    strat_train_set = df.loc[train_index]
    strat_test_set = df.loc[test_index]
```

```python
strat_train_set
```

```python
strat_train_set['Income_Category'].value_counts().sort_index().plot.bar(rot = 0, grid = True)
```

```python
strat_test_set
```

```python
strat_test_set['Income_Category'].value_counts().sort_index().plot.bar(rot = 0, grid = True)
```

```python
for dataframe in (strat_train_set, strat_test_set):
    dataframe.drop('Income_Category', axis = 1, inplace = True)
```

```python
df = strat_train_set.copy()
```

```python
house_labels = df['median_house_value'].copy()
housing = df.drop('median_house_value', axis = 1)
```

```python
house_labels
```

```python
housing
```

```python
housing.info()
```

```python
#n As we can see there are missing values in total_bedrooms column

from sklearn.impute import SimpleImputer
```

```python
imputer = SimpleImputer(strategy='mean')
housing_num = housing.select_dtypes(include=[np.number])
imputer.fit(housing_num)
```

```python
imputer.statistics_
```

```python
housing_num.info()
```

```python
dataf = imputer.transform(housing_num)
```

```python
np.isnan(dataf).sum()
```

```python
np.isnan(dataf).any()
```

```python
'''
Other available strategies:

"median" – replaces with the middle value of a dataset

"most_frequent" – for the most common value (can handle categorical)

"constant" – fill with a fixed value using fill_value=... like SimpleImputer(strategy='constant', fill_value = 'xyz')
'''
```

```python
Final_Training_Data = pd.DataFrame(dataf, columns = housing_num.columns, index = housing_num.index)
```
